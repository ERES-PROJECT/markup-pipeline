```python
def pjm_tf_prepare(df):
    """Prepare monthly all-years RT PJM data for database import."""

    df = df.copy()

    df['date'] = pd.to_datetime(
        dict(
            year=df['Year'],
            month=df['Month'],
            day=1
        )
    ).dt.date

    df['zone'] = df['EDC'].astype('string')
    df['average_lmp'] = pd.to_numeric(df['Average'], errors='coerce')

    df['import_id'] = None

    df = df[
        ['date', 'zone', 'average_lmp', 'import_id']
    ]

    df = df.where(pd.notnull(df), None)

    return df

```
